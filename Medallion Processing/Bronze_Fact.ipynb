```python
from pyspark.sql import functions as F

volume_path = "/Volumes/ecommerce/bronze/rawdata/order_items/landing/"

df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(volume_path + "*.csv")
)

display(df.limit(10))
```

```python
df.write.format("delta")\
    .mode("overwrite")\
    .option("overwriteSchema", "true")\
    .saveAsTable("ecommerce.bronze.brz_order_items")
```
